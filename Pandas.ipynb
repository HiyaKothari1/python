```python
import pandas as pd

df = pd.DataFrame({
    "country": ["USA","USA","India","India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22]
})
melted_df = df.melt(
    id_vars = ["country","year"],
    value_vars= ["sales","profit"],
    var_name="metrics",
    value_name="value"
)
original = melted_df.pivot(
    index=["country","year"],
    columns="metrics",
    values="value"
)

```

```python
#web scraping activity 1
import requests
from bs4 import BeautifulSoup

```

```python
#Fetch data and store it
URL = "https://quotes.toscrape.com/page/1/"
page_count = 1
while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL) #sends an HTTP request to the website

    soup = BeautifulSoup(res.text,"lxml")#parses html contnet
    quotes = soup.select("div.quote")#finds all quote containers

    if not quotes:
        print("no valid pages anymore.. ")
        break #stops the loop when no quotes are found 

    with open(f"scrapped_data/quotes{page_count}.html","w" , encoding="utf-8") as f: #creates and opens an html file 
        f.write(res.text) #saves html content 
        print(f"downloaded data from page{page_count}")
    page_count = page_count +1

 
```

```python

from bs4 import BeautifulSoup

life_quotes = []

for page_count in range(1, 11):

    with open(f"scrapped_data/quotes{page_count}.html", "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = []

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()

            life_quotes.append([text, author])

print(life_quotes)
print("Total life quotes:", len(life_quotes))
```
